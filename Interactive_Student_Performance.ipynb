```python
import pandas as pd
import plotly.express as px
data = pd.read_csv("student_performance_5000_rows.csv")
fig = px.scatter(
    data,
    x="Study Hours",
    y="Marks",
    title="Interactive Scatter Plot: Study Hours vs Marks",
    hover_data=["Attendance"],
    opacity=0.6
)

fig.update_layout(
    xaxis_title="Study Hours",
    yaxis_title="Marks",
    template="plotly_white"
)

fig.show()
```

```python
import pandas as pd
import plotly.express as px
data = pd.read_csv("student_performance_5000_rows.csv")
data_sorted = data.sort_values("Attendance")
fig = px.line(
    data_sorted,
    x="Attendance",
    y="Marks",
    markers=True,
    title="Interactive Line Plot: Attendance vs Marks"
)

fig.update_layout(
    xaxis_title="Attendance (%)",
    yaxis_title="Marks",
    template="plotly_white"
)

fig.show()
```
